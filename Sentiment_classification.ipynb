```python
!pip install -q transformers torch scikit-learn wandb pandas matplotlib tqdm
import torch
from transformers import DistilBertTokenizer, DistilBertForSequenceClassification
from torch.optim import AdamW
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import pandas as pd
import wandb
from torch.utils.data import Dataset, DataLoader
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
from google.colab import drive
import numpy as np


# 1. Data loading and preparation
def load_and_prepare_data():
    drive.mount('/content/drive')
    df = pd.read_csv('/content/drive/My Drive/7LLMO/tripadvisor_hotel_reviews.csv')
    df['sentiment'] = df['Rating'].apply(lambda x: 0 if x <= 2 else 1 if x == 3 else 2)
    print("Sentiment distribution:\n", df['sentiment'].value_counts())
    return train_test_split(df, test_size=0.15, random_state=42)

# 2. Model and tokenizer
def initialize_model():
    tokenizer = DistilBertTokenizer.from_pretrained("distilbert-base-uncased")
    model = DistilBertForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=3)
    print("\nModel initialized")
    return tokenizer, model

# 3. Dataset class
class ReviewDataset(Dataset):
    def __init__(self, texts, labels, tokenizer):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = labels

    def __getitem__(self, idx):
        return {**{k: v[idx] for k, v in self.encodings.items()}, 'labels': torch.tensor(self.labels[idx])}

    def __len__(self):
        return len(self.labels)

# 4. Training loop
def train_model(model, train_loader, val_loader, epochs=3, lr=3e-5):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)
    optimizer = AdamW(model.parameters(), lr=lr)

    # Wandb initialization
    wandb.init(project="fast-sentiment-3class", config={
        "epochs": epochs,
        "learning_rate": lr,
        "batch_size": 32
    })

    for epoch in range(epochs):
        model.train()
        total_loss = 0
        for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}"):
            optimizer.zero_grad()
            batch = {k: v.to(device) for k, v in batch.items()}
            outputs = model(**batch)
            loss = outputs.loss
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        # Validation
        model.eval()
        val_loss, correct = 0, 0
        with torch.no_grad():
            for batch in tqdm(val_loader, desc="Validating"):
                batch = {k: v.to(device) for k, v in batch.items()}
                outputs = model(**batch)
                val_loss += outputs.loss.item()
                correct += (outputs.logits.argmax(1) == batch['labels']).sum().item()

        # Logging metrics
        wandb.log({
            'epoch': epoch + 1,
            'train_loss': total_loss/len(train_loader),
            'val_loss': val_loss/len(val_loader),
            'val_acc': correct/len(val_loader.dataset)
        })
        print(f"\nEpoch {epoch+1}/{epochs}: Val Acc: {correct/len(val_loader.dataset):.4f}")

    wandb.finish()

# 5. Visualization
def visualize_results():
    api = wandb.Api()
    runs = api.runs("simon-kunz/fast-sentiment-3class")
    if runs:
        run = sorted(runs, key=lambda x: x.created_at)[-1]
        history = run.history()

        print(f"\nFound {len(history)} logged epochs in W&B")

        plt.figure(figsize=(12, 5))

        # Graph loss
        plt.subplot(1, 2, 1)
        plt.plot(history['epoch'], history['train_loss'], 'b-', label='Train Loss')
        plt.plot(history['epoch'], history['val_loss'], 'r-', label='Val Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.title('Training and Validation Loss')
        plt.legend()
        plt.grid(True)

        # Graph accuracy
        plt.subplot(1, 2, 2)
        plt.plot(history['epoch'], history['val_acc'], 'g-', label='Val Accuracy')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.title('Validation Accuracy')
        plt.legend()
        plt.grid(True)

        plt.tight_layout()
        plt.show()
    else:
        print("No W&B runs found")

# 6. Inference on some examples
def predict_examples(model, tokenizer, device):
    examples = [
        "The hotel was absolutely disgusting!",
        "It was okay, nothing special.",
        "Loved every minute of our stay! Perfect service!",
        "Our stay at the accomodation was average.",
        "It was not anything spectacular."
    ]

    model.eval()
    print("\n" + "="*50)
    print("EXAMPLE PREDICTIONS:".center(50))
    print("="*50)
    for text in examples:
        inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=64).to(device)
        with torch.no_grad():
            outputs = model(**inputs)
            probs = torch.nn.functional.softmax(outputs.logits, dim=1)[0]
            pred = ['NEGATIVE', 'NEUTRAL', 'POSITIVE'][outputs.logits.argmax()]

        print("\n" + "-"*50)
        print(f"TEXT: {text}")
        print(f"PREDICTION: {pred}")
        print(f"PROBABILITIES: neg={probs[0]:.3f}, neut={probs[1]:.3f}, pos={probs[2]:.3f}")
    print("\n" + "="*50)

# Main
def main():
    # 1. Data
    print("Loading and preparing data...")
    train_df, test_df = load_and_prepare_data()
    train_df, val_df = train_test_split(train_df, test_size=0.176, random_state=42)

    print("\nData splits:")
    print(f"- Training samples: {len(train_df)}")
    print(f"- Validation samples: {len(val_df)}")
    print(f"- Test samples: {len(test_df)}")

    # 2. Model
    print("\nInitializing model...")
    tokenizer, model = initialize_model()

    # 3. Datasets
    print("\nCreating datasets...")
    train_dataset = ReviewDataset(train_df['Review'].tolist(), train_df['sentiment'].tolist(), tokenizer)
    val_dataset = ReviewDataset(val_df['Review'].tolist(), val_df['sentiment'].tolist(), tokenizer)
    test_dataset = ReviewDataset(test_df['Review'].tolist(), test_df['sentiment'].tolist(), tokenizer)

    # 4. Model training
    print("\nStarting training...")
    train_model(model,
               DataLoader(train_dataset, batch_size=32, shuffle=True),
               DataLoader(val_dataset, batch_size=32),
               epochs=3)

    # 5. Evaluation
    print("\nEvaluating on test set...")
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for batch in tqdm(DataLoader(test_dataset, batch_size=32), desc="Testing"):
            batch = {k: v.to(device) for k, v in batch.items()}
            all_preds.extend(model(**batch).logits.argmax(1).cpu().numpy())
            all_labels.extend(batch['labels'].cpu().numpy())

    print("\nTEST SET RESULTS:")
    print(classification_report(all_labels, all_preds,
                               target_names=['negative', 'neutral', 'positive'],
                               digits=4))

    # 6. Visualisation
    print("\nGenerating visualizations...")
    visualize_results()

    # 7. Example predictions
    predict_examples(model, tokenizer, device)

    # Saving
    print("\nSaving model to Google Drive...")
    model_path = "/content/drive/MyDrive/7LLMO/sentiment_model"
    !mkdir -p "/content/drive/MyDrive/7LLMO"
    try:
        model.save_pretrained(model_path)
        tokenizer.save_pretrained(model_path)
        print(f"Model successfully saved in: {model_path}")

    except Exception as e:
        print(f"Error during saving the model: {e}")

if __name__ == "__main__":
    main()
```

```python
# Inference
from transformers import DistilBertTokenizer, DistilBertForSequenceClassification
import torch
from google.colab import drive
import os

def run_inference():
    drive.mount('/content/drive', force_remount=False)


    model_dir = "/content/drive/MyDrive/7LLMO/sentiment_model"


    if not os.path.exists(model_dir):
        print(f"Model directory was not found: {model_dir}")
        return


    try:
        print("Loading model...")
        tokenizer = DistilBertTokenizer.from_pretrained(model_dir)
        model = DistilBertForSequenceClassification.from_pretrained(model_dir)
        model.eval().to('cuda' if torch.cuda.is_available() else 'cpu')
        print("Model was successfully loaded")
    except Exception as e:
        print(f"Error happened during fetching model: {e}")
        return

    # 5. Input inference
    print("\nTRY INFERENCE (type 'q' to quit)")
    while True:
        text = input("\nEnter text for sentiment analysis: ")
        if text.lower() == 'q':
            break

        if text.strip():
            try:
                inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=64).to(model.device)
                with torch.no_grad():
                    outputs = model(**inputs)
                    probs = torch.nn.functional.softmax(outputs.logits, dim=1)[0]

                print("\Result:")
                print(f"Prediction: {['NEGATIVE', 'NEUTRAL', 'POSITIVE'][probs.argmax()]}")
                print(f"Probability: Negative={probs[0]:.2f}, Neutral={probs[1]:.2f}, Positive={probs[2]:.2f}")
            except Exception as e:
                print(f"Error happened during analysis: {e}")
        else:
            print("Please enter valid text!")

run_inference()
```
